# TrainPilot on one free GPU

Set the runtime to a single NVIDIA T4, then run the cell below.
The cell clones this repo if needed, runs a short check of the three reference setups, and does not edit those scripts.
The same cell works in Colab and Kaggle.


```python
import os
import runpy
import subprocess
import sys
from pathlib import Path

parent = Path("/content") if Path("/content").exists() else Path("/kaggle/working")
root = parent / "TrainPilot"
if not (root / "notebooks" / "run_t4_checks.py").exists():
    subprocess.check_call(["git", "clone", "https://github.com/linee64/TrainPilot.git", str(root)])
os.chdir(root)
if (root / ".git").exists():
    subprocess.check_call(["git", "pull"])
subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "--upgrade", "--force-reinstall", "--no-deps", "."])
subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "datasets"])
for name in list(sys.modules):
    if name == "trainpilot" or name.startswith("trainpilot."):
        del sys.modules[name]
runpy.run_path(str(root / "notebooks" / "run_t4_checks.py"), run_name="__main__")
```
